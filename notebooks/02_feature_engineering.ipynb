```python
import pandas as pd
df = pd.read_csv("../data/processed/sfax_weather_cleaned.csv", parse_dates=['valid_time'])

df['hour'] = df['valid_time'].dt.hour
df['day'] = df['valid_time'].dt.day
df['dayofweek'] = df['valid_time'].dt.dayofweek
df['month'] = df['valid_time'].dt.month
df['year'] = df['valid_time'].dt.year
df['is_weekend'] = df['dayofweek'].isin([5, 6]).astype(int)

df['t2m_lag_1'] = df['t2m'].shift(1)
df['t2m_lag_24'] = df['t2m'].shift(24)
df['tp_lag_1'] = df['tp'].shift(1)
df['tp_lag_24'] = df['tp'].shift(24)

df['t2m_rollmean_3'] = df['t2m'].rolling(window=3).mean()
df['t2m_rollmean_24'] = df['t2m'].rolling(window=24).mean()

df = df.dropna().reset_index(drop=True)

processed_path = "../data/processed/sfax_weather_features.csv"
df.to_csv(processed_path, index=False)
print(f"Feature-engineered data saved to {processed_path}")

display(df.head())
print(df.shape)
```
