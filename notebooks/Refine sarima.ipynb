```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from pmdarima import auto_arima
from sklearn.metrics import mean_absolute_error, mean_squared_error

df = pd.read_csv("../data/processed/sfax_weather_features.csv", parse_dates=['valid_time'])
df = df.set_index('valid_time')

train = df.loc[:'2023-12-31']['t2m']
val = df.loc['2024-01-01':'2024-06-30']['t2m']

stepwise_model = auto_arima(train, seasonal=True, m=24, trace=True, suppress_warnings=True)

forecast = stepwise_model.predict(n_periods=len(val))

plt.figure(figsize=(15,5))
plt.plot(train[-500:], label='Train')
plt.plot(val, label='Validation')
plt.plot(val.index, forecast, label='Forecast (auto_arima)')
plt.xlabel('Date')
plt.ylabel('Temperature (°C)')
plt.title('Hourly SARIMA (auto_arima) Forecast vs Actual (t2m)')
plt.legend()
plt.show()

mae = mean_absolute_error(val, forecast)
rmse = np.sqrt(mean_squared_error(val, forecast))
print(f"MAE: {mae}")
print(f"RMSE: {rmse}")
```
